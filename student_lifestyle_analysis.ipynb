```python
import pandas as pd
```

```python
df = pd.read_excel("Student Lifestyle Survey  (Responses).xlsx")
```

```python
df.head()
```

```python
clean_df = df.copy()
```

```python
print("Rows:", clean_df.shape[0])
print("Columns:", clean_df.shape[1])
```

```python
clean_df.isnull().sum()
```

```python
clean_df.duplicated().sum()
```

```python
clean_df = clean_df.drop(columns=["Timestamp"])
```

```python
column_names = {

    "Which class are you in?": "class",

    "What is your gender?": "gender",

    "How long have you been preparing for NEET?": "neet_prep_duration",

    "Do you take any additional coaching outside this institute?": "additional_coaching",

    "On a typical weekday, how many hours do you study (self-study only, not coaching)?": "self_study_hours",

    "Do you study on weekends?": "weekend_study",

    "Which subject do you find hardest?": "hardest_subject",

    "How many mock tests do you attempt per month?": "mock_tests_per_month",

    "Do you review your mistakes after each mock test?": "mistake_review",

    "On average, how many hours do you sleep per night?\n": "sleep_hours",

    "What time do you usually go to sleep?": "sleep_time",

    "How many hours per day do you spend on your phone (non-study)?": "phone_hours",

    "Do you use your phone within 30 minutes of sleeping?": "phone_before_sleep",

    "How stressed do you feel about NEET preparation on most days?": "stress_level",

    "How confident are you about clearing NEET this year?": "confidence_level",

    "Do you take any breaks or do physical activity during the day?": "physical_activity",

    "Do you do any other extra activities?(hobbies like gym, sports, music, art,  dance, writing)": "extra_activities",

    "How much time do u spend on those Activity?": "activity_hours",

    "How would you describe your current mental state during preparation?": "mental_state",

    "What was your score (%) in the most recent mock test you attempted?": "mock_score_range",

    "Compared to 3 months ago, your mock test scores are:": "score_trend",

    "What is your target NEET score?": "target_score_range"
}

clean_df = clean_df.rename(columns=column_names)

print(clean_df.columns.tolist())
```

```python
clean_df["hardest_subject"].value_counts()
```

```python
clean_df["hardest_subject"] = (
    clean_df["hardest_subject"]
    .str.strip()
    .str.lower()
    .replace({
        "phy": "Physics",
        "physics": "Physics",
        "chem": "Chemistry",
        "chemistry": "Chemistry",
        "math": "Mathematics",
        "mathematics": "Mathematics",
        "bio": "Biology",
        "biology": "Biology"
    })
)
```

```python
clean_df["hardest_subject"].value_counts()
```

```python
clean_df["physical_activity"].value_counts()
```

```python
clean_df["physical_activity"] = (
    clean_df["physical_activity"]
    .str.strip()
    .replace({
        "Yes, regularlt": "Yes, regularly",
        "regularlt": "Yes, regularly",
        "rarely": "Rarely",
        "occasionally": "Occasionally"
    })
)
```

```python
clean_df["physical_activity"].value_counts()
```

```python
clean_df["weekend_study"] = (
    clean_df["weekend_study"]
    .str.strip()
    .replace({
        "Yes, Same as weekedays": "Yes, same as weekdays"
    })
)
```

```python
clean_df["score_trend"] = (
    clean_df["score_trend"]
    .str.strip()
    .str.capitalize()
)
```

```python
clean_df["self_study_hours"].value_counts()
```

```python
study_hours_map = {
    "Less than 2hrs": 1,
    "2-4 hrs": 3,
    "4-6hrs": 5,
    "More than 6 hrs": 7
}

clean_df["self_study_hours_mid"] = (
    clean_df["self_study_hours"]
    .map(study_hours_map)
)
```

```python
sleep_hours_map = {
    "Less than 5hrs": 4,
    "5-6hrs": 5.5,
    "6-7hrs": 6.5,
    "7-8hrs": 7.5,
    "More than 8hrs": 9
}

clean_df["avg_sleep_hours"] = (
    clean_df["sleep_hours"]
    .map(sleep_hours_map)
)
```

```python
phone_hours_map = {
    "Less than 1hr": 0.5,
    "1-2hrs": 1.5,
    "2-4hrs": 3,
    "More than 4hrs": 5
}

clean_df["avg_phone_hours"] = (
    clean_df["phone_hours"]
    .map(phone_hours_map)
)
```

```python
activity_hours_map = {
    "Less than 1 hr": 0.5,
    "1-2hrs": 1.5,
    "More than 2hrs": 3
}

clean_df["avg_activity_hours"] = (
    clean_df["activity_hours"]
    .map(activity_hours_map)
)
```

```python
mock_score_map = {
    "Below 30%": 20,
    "30-50%": 40,
    "50-70%": 60,
    "70-85%": 77.5,
    "Above 85%": 90
}

clean_df["mock_score_mid"] = (
    clean_df["mock_score_range"]
    .map(mock_score_map)
)
```

```python
target_score_map = {
    "500-600": 550,
    "600-650": 625,
    "650-700": 675,
    "700+": 725
}

clean_df["target_score_mid"] = (
    clean_df["target_score_range"]
    .map(target_score_map)
)
```
